Primero, usamos la clase Scanner para poder escribir y leer los datos que el usuario introduce por la consola, como la cantidad total de alumnos y cuántos van en cada grupo.

Después, guardamos esos dos números ingresados en variables enteras (int totalEstudiantes y int estudiantesPorEquipo), tal como nos pedía la consigna.

Luego, el programa imprime en pantalla los valores que acabamos de ingresar para confirmar que todo esté bien.

Para saber cuántos grupos armamos, usamos una división entera (/), ya que al dividir el total entre los integrantes por equipo, Java nos da directamente la cantidad de equipos completos.

Para calcular cuántos estudiantes quedan sueltos o sin grupo, usamos el operador módulo (%), que básicamente nos devuelve el resto de esa división.

Finalmente, mostramos en la consola los mensajes con los resultados claros y cerramos el Scanner para terminar. ¡Y listo, con eso el programa cumple con todo!


```python
import java.util.Scanner;

public class DistribucionEstudiantes {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);

        // Solicitar y leer la cantidad total de estudiantes
        System.out.print("Ingrese la cantidad total de estudiantes: ");
        int totalEstudiantes = scanner.nextInt();

        // Solicitar y leer la cantidad de estudiantes por equipo
        System.out.print("Ingrese la cantidad de estudiantes por equipo: ");
        int estudiantesPorEquipo = scanner.nextInt();

        // Mostrar los valores ingresados
        System.out.println();
        System.out.println("Cantidad total de estudiantes: " + totalEstudiantes);
        System.out.println("Estudiantes por equipo: " + estudiantesPorEquipo);

        // Calcular equipos completos y sobrantes
        int equiposCompletos = totalEstudiantes / estudiantesPorEquipo;
        int estudiantesRestantes = totalEstudiantes % estudiantesPorEquipo;

        // Mostrar los resultados
        System.out.println("Se pueden formar " + equiposCompletos + " equipos completos.");
        System.out.println("Quedan " + estudiantesRestantes + " estudiantes sin integrar un equipo completo.");

        scanner.close();
    }
}
```
