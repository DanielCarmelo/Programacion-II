```python
import java.util.Random;

public class DiaDeLaSemana {
    public static void main(String[] args) {
        // Crear objeto de la clase Random para generar números aleatorios
        Random random = new Random();

        // Generar un número entero aleatorio entre 1 y 7
        // random.nextInt(7) genera valores de 0 a 6; sumando 1 obtenemos el rango de 1 a 7
        int numeroGenerado = random.nextInt(7) + 1;

        // Mostrar el número generado
        System.out.println("Número generado: " + numeroGenerado);

        // Estructura switch para determinar el día de la semana
        String diaSemana;
        switch (numeroGenerado) {
            case 1:
                diaSemana = "lunes";
                break;
            case 2:
                diaSemana = "martes";
                break;
            case 3:
                diaSemana = "miércoles";
                break;
            case 4:
                diaSemana = "jueves";
                break;
            case 5:
                diaSemana = "viernes";
                break;
            case 6:
                diaSemana = "sábado";
                break;
            case 7:
                diaSemana = "domingo";
                break;
            default:
                diaSemana = "Número inválido";
                break;
        }

        // Mostrar el día correspondiente
        System.out.println("Día correspondiente: " + diaSemana);
    }
}
```

# Explicación del Programa: Día de la Semana en Java

Este programa genera un número entero de forma aleatoria y utiliza una estructura de control condicional para asociar dicho número con el día correspondiente de la semana.

   Usa la clase `java.util.Random`, encargada de proveer métodos para generar números pseudoaleatorios.

2. **Generación del número entero (1 a 7)**
   * Utiliza la función `random.nextInt(7)` la cual produce números enteros en el intervalo de **0 a 6**.
   * Le suma `1` al resultado para desplazar el rango y obtener exactamente números entre **1 y 7**.

3. **Impresión del número generado**
   Muestra en la consola el valor aleatorio obtenido para que el usuario sepa qué número evaluará el programa.

4. **Evaluación condicional mediante `switch`**
   Compara el valor de la variable `numeroGenerado` contra cada uno de los casos (`case`):
   * `1` ➔ `"lunes"`
   * `2` ➔ `"martes"`
   * `3` ➔ `"miércoles"`
   * `4` ➔ `"jueves"`
   * `5` ➔ `"viernes"`
   * `6` ➔ `"sábado"`
   * `7` ➔ `"domingo"`
   
   > **Nota:** Cada caso cuenta con una instrucción `break` para detener la ejecución de la estructura `switch` una vez hallada la coincidencia. También incluye un bloque `default` como medida de seguridad ante valores fuera de rango.
