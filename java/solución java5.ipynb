```python
public class ContarVocales {
    public static void main(String[] args) {
        // Variable String con el texto a analizar
        String texto = "PROGRAMACION JAVA";

        // Variable contadora de vocales
        int cantidadVocales = 0;

        // Recorrido de los caracteres del texto mediante un ciclo for
        for (int i = 0; i < texto.length(); i++) {
            char caracter = texto.charAt(i);

            // Verificar si el carácter es una vocal (mayúscula o minúscula)
            if (caracter == 'A' || caracter == 'E' || caracter == 'I' || caracter == 'O' || caracter == 'U' ||
                caracter == 'a' || caracter == 'e' || caracter == 'i' || caracter == 'o' || caracter == 'u') {
                cantidadVocales++;
            }
        }

        // Mostrar los resultados requeridos
        System.out.println("Texto analizado: " + texto);
        System.out.println("Cantidad de caracteres: " + texto.length());
        System.out.println("Cantidad de vocales: " + cantidadVocales);
    }
}
```

# Explicación del Programa: Cantidad de Vocales en un Texto

Este programa analiza una cadena de texto predefinida, cuenta la cantidad total de caracteres que posee y determina cuántas vocales hay en ella mediante un bucle de repetición.

## 🛠️ ¿Qué hace el programa paso a paso?

1. **Declaración de variables**
   * `texto`: Variable de tipo `String` que almacena la frase `"PROGRAMACION JAVA"`.
   * `cantidadVocales`: Variable entera inicializada en `0`, que actúa como contador.

2. **Estructura repetitiva (`for`)**
   * El ciclo se ejecuta desde el índice `0` hasta `texto.length() - 1`, recorriendo cada posición del texto.

3. **Extracción y evaluación de caracteres**
   * En cada iteración, se obtiene el carácter actual usando el método `texto.charAt(i)`.
   * Una estructura condicional `if` verifica si dicho carácter coincide con alguna vocal (evaluando tanto mayúsculas como minúsculas).
   * Si la condición se cumple, incrementa la variable `cantidadVocales` en 1 (`cantidadVocales++`).

4. **Despliegue de resultados**
   * Imprime el contenido de la variable `texto`.
